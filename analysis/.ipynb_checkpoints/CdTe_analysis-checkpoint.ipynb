```python
import pandas as pd
from matplotlib import pyplot
import seaborn as sns
```

```python
df_cirrus = pd.read_csv('../csv_data/CdTe_Cirrus_combined.csv')
df_cirrus['LOOP+ per Band'] = df_cirrus['LOOP+ Time'] / df_cirrus['Bands']
df_cirrus['Performance'] = 1 / df_cirrus['LOOP+ per Band']
df_cirrus['Cores per Node'] = df_cirrus['Cores'] / df_cirrus['Nodes']
print(df_cirrus['Cores per Node'].unique())
```

```python
df_cirrus.head()
```

```python
df_cirrus_mpi = df_cirrus.loc[df_cirrus['Threads'] == 1]
df_cirrus_full = df_cirrus_mpi.loc[df_cirrus_mpi['Cores per Node'] == 288]
df_cirrus_single = df_cirrus_mpi.loc[df_cirrus_mpi['Nodes'] == 1]
```

```python
sns.lineplot(data=df_cirrus_full, x='Cores', y='Performance', hue='Label', err_style='bars')
```

```python
sns.lineplot(data=df_cirrus_full, x='Cores', y='Energy', hue='Label', err_style='bars')
```

```python
sns.barplot(data=df_cirrus_single, x='Cores per Node', y='Performance', hue='Label')
```
